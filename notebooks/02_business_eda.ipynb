```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('ecommerce_cleaned.csv')
```

```python
total_revenue = df['revenue'].sum()

print(
    f'Total Revenue: {total_revenue:,.2f}'
)
```

```python
total_orders = df['order_id'].nunique()

print(
    f'Total Orders: {total_orders}'
)
```

```python
AOV = total_revenue / total_orders

print(
    f'Average Order Value: {AOV:,.2f}'
)
```

```python
monthly_revenue = (

    df.groupby('purchase_month')

    ['revenue']

    .sum()

    .reset_index()

)
```

```python
plt.figure(figsize=(15,6))

sns.lineplot(

    data=monthly_revenue,

    x='purchase_month',

    y='revenue'

)

plt.title('Monthly Revenue Trend')

plt.xticks(rotation=45)

plt.show()
```

### Insight

Revenue increased significantly during Q4,
indicating strong seasonal shopping behavior.

```python
top_category = (

    df.groupby(
        'product_category_name_english'
    )

    ['revenue']

    .sum()

    .sort_values(ascending=False)

    .head(10)

)
```

```python
plt.figure(figsize=(12,6))

top_category.plot(kind='bar')

plt.title('Top Product Categories')

plt.show()
```

```python
review_analysis = (

    df.groupby('review_score')

    ['delivery_days']

    .mean()

)
```

```python
plt.figure(figsize=(8,5))

review_analysis.plot(kind='bar')

plt.title('Review Score vs Delivery Days')

plt.show()
```

### Insight

Orders with longer delivery times tend to receive lower review scores.

This indicates that delivery performance directly impacts customer satisfaction.

```python
customer_state = (

    df.groupby('customer_state')

    ['revenue']

    .sum()

    .sort_values(ascending=False)

)
```

```python
plt.figure(figsize=(12,6))

customer_state.head(10).plot(kind='bar')

plt.title('Top Revenue by State')

plt.show()
```
